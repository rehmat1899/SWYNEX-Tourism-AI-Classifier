```python

questions = [
    "What are the famous places in Murree?",
    "Which attractions should I visit in Hunza?",
    "How can I reach Naran from Islamabad?",
    "What is the best route to Swat?",
    "Where can I find hotels in Murree?",
    "Is there affordable accommodation in Hunza?",
    "What is the weather like in Naran?",
    "Will it rain in Swat tomorrow?",
    "Is it safe to travel to northern areas?",
    "Are the roads safe for tourists?"
]

categories = [
    "Attractions",
    "Attractions",
    "Routes",
    "Routes",
    "Accommodation",
    "Accommodation",
    "Weather",
    "Weather",
    "Safety",
    "Safety"
]

print("Total Questions:", len(questions))
print("Total Categories:", len(categories))
```

```python

import pandas as pd

data = {
    "question": [
        # Attractions (10)
        "What are the famous places in Murree?",
        "Which attractions should I visit in Hunza?",
        "What is famous in Skardu?",
        "Are there beautiful lakes in Naran?",
        "Which historical places can I visit in Lahore?",
        "What are the best tourist spots in Kashmir?",
        "Where can I see mountains in Gilgit?",
        "Which parks are popular in Islamabad?",
        "What are the famous valleys in Pakistan?",
        "Which museums can I visit in Lahore?",

        # Routes (10)
        "How can I reach Naran from Islamabad?",
        "What is the best route to Swat?",
        "How do I travel from Lahore to Murree?",
        "Which road leads to Hunza?",
        "How can I get to Skardu?",
        "What is the distance from Islamabad to Naran?",
        "Are there buses from Rawalpindi to Murree?",
        "How can I travel to Neelum Valley?",
        "Which route goes to Fairy Meadows?",
        "How do I reach Gilgit from Islamabad?",

        # Accommodation (10)
        "Where can I find hotels in Murree?",
        "Is there affordable accommodation in Hunza?",
        "Which hotels are available in Skardu?",
        "Where can tourists stay in Naran?",
        "Are guest houses available in Swat?",
        "What is the hotel price in Murree?",
        "Can I book a room in Gilgit?",
        "Where are the best resorts in Kashmir?",
        "Is camping accommodation available in Naran?",
        "Which hotels provide family rooms in Hunza?",

        # Weather (10)
        "What is the weather like in Naran?",
        "Will it rain in Swat tomorrow?",
        "Is it snowing in Murree?",
        "What is the temperature in Hunza?",
        "Will the weather be cold in Skardu?",
        "What is the weather forecast for Gilgit?",
        "Is it sunny in Islamabad today?",
        "When does snowfall start in Murree?",
        "Is there a chance of rain in Kashmir?",
        "What is the temperature in Naran today?",

        # Safety (10)
        "Is it safe to travel to northern areas?",
        "Are the roads safe for tourists?",
        "Is Hunza safe for families?",
        "Are there any travel safety warnings for Naran?",
        "Is it safe to visit Kashmir?",
        "What precautions should tourists take in Skardu?",
        "Is travelling to Swat safe?",
        "Are mountain roads dangerous during rain?",
        "Is it safe to travel at night in Murree?",
        "Are there emergency services in Hunza?"
    ],

    "category": (
        ["Attractions"] * 10 +
        ["Routes"] * 10 +
        ["Accommodation"] * 10 +
        ["Weather"] * 10 +
        ["Safety"] * 10
    )
}

df = pd.DataFrame(data)

print("Dataset created successfully!")
print("Total questions:", len(df))
print("\nQuestions in each category:")
print(df["category"].value_counts())

df.head()
```

```python

# Import machine learning tools

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Divide dataset into training and testing data

X_train, X_test, y_train, y_test = train_test_split(
    df["question"],
    df["category"],
    test_size=0.2,
    random_state=42,
    stratify=df["category"]
)

# Convert text into numerical features

vectorizer = TfidfVectorizer()
X_train_vectorized = vectorizer.fit_transform(X_train)
X_test_vectorized = vectorizer.transform(X_test)

# Create and train the AI model

model = MultinomialNB()
model.fit(X_train_vectorized, y_train)

# Test the model

predictions = model.predict(X_test_vectorized)

accuracy = accuracy_score(y_test, predictions)

print("AI Model trained successfully!")
print("Training questions:", len(X_train))
print("Testing questions:", len(X_test))
print("Model Accuracy:", round(accuracy * 100, 2), "%")
```

```python

def classify_question(question):

    # Convert the question into numerical features
    question_vector = vectorizer.transform([question])

    # Predict its category
    prediction = model.predict(question_vector)

    return prediction[0]


# Test our AI model

test_questions = [
    "What is the weather in Murree?",
    "Where can I find hotels in Hunza?",
    "How can I reach Naran?",
    "What are famous places in Skardu?",
    "Is it safe to travel to Swat?"
]

for question in test_questions:
    result = classify_question(question)

    print("Question:", question)
    print("Predicted Category:", result)
    print("-----------------------------")
```

```python

from sklearn.metrics import classification_report, confusion_matrix

print("MODEL PERFORMANCE REPORT")
print("------------------------")

print(classification_report(y_test, predictions))

print("CONFUSION MATRIX:")
print(confusion_matrix(y_test, predictions))
```

```python
print("SMART TOURISM AI CLASSIFIER")
print("----------------------------")

while True:
    user_question = input("Enter your tourism question (or type exit): ")

    if user_question.lower() == "exit":
        print("Thank you for using our AI model!")
        break

    result = classify_question(user_question)

    print("Predicted Category:", result)
    print("----------------------------")
```

```python
import joblib

joblib.dump(model, "tourism_model.pkl")
joblib.dump(vectorizer, "tfidf_vectorizer.pkl")

print("Model saved successfully!")
print("Vectorizer saved successfully!")
```
